```python
import numpy as np

# 1. Create a 1D NumPy Array
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
```

```python
# 2. Inspection and Properties
print("--- Week 1 Array Properties ---")
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

```python
# 3. Array Operations
fahrenheit_w1 = (temperatures_w1 * 9/5) + 32
print("\n--- Operations ---")
print("Temperatures in Fahrenheit:", fahrenheit_w1)
print("Maximum Temperature:", temperatures_w1.max())
print("Minimum Temperature:", temperatures_w1.min())
print("Mean Temperature:", temperatures_w1.mean())
```

```python
# 4. Array Slicing and Indexing
print("\n--- Slicing 1D Array ---")
print("First three days:", temperatures_w1[:3])
print("Weekend (last two days):", temperatures_w1[-2:])
print("Middle three days:", temperatures_w1[2:5])
```

```python
import pandas as pd

# 1. Creating Pandas Series
marks = pd.Series([95, 92, 89, 85, 80], index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
```

```python
# 2. Indexing and Slicing
rank1_mark = marks.iloc[0]                          # Integer index position for 1st rank
top_3_ranks = marks.loc[['Rank1', 'Rank2', 'Rank3']] # Access via index labels
rank3_mark = marks.iloc[2]                           # Integer index position for 3rd rank
filtered_ranks = marks[marks > 90]                   # Boolean mask filtering
print("--- Series Indexing ---")
print(f"1st Rank Mark: {rank1_mark}")
print(f"Top 3 Ranks:\n{top_3_ranks}")
print(f"3rd Rank Mark: {rank3_mark}")
print(f"Ranks with marks > 90:\n{filtered_ranks}")
```

```python
# 3. Manipulating Series
marks['Rank1'] = 100                                 # Modify 1st rank to 100
marks = marks.drop('Rank5')                          # Remove the entry corresponding to the last rank
cgpa = marks / 10                                    # Compute CGPA

print("\n--- Manipulated Series & CGPA ---")
print(cgpa)
```

```python
import pandas as pd

# 1. Creating Pandas DataFrame
data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Furniture', 'Electronics'],
    'Region': ['North', 'South', 'North', 'East', 'West', 'North', 'East', 'West', 'South', 'North'],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}
transactions = pd.DataFrame(data)
```

```python
# 2. Data Exploration
print("--- Head ---")
print(transactions.head())
print("\n--- Tail ---")
print(transactions.tail())
print("\n--- Info Summary ---")
print(transactions.info())  # Evaluates shape, column names, and data types

print("\n--- Select Columns ---")
print(transactions[['ProductCategory', 'Amount']])

print("\n--- Last 3 Columns ---")
print(transactions.iloc[:, -3:])

print("\n--- Filtered Rows (Region=='North' & Amount > 200) ---")
print(transactions[(transactions['Region'] == 'North') & (transactions['Amount'] > 200)])

print("\n--- Value Counts (ProductCategory) ---")
print(transactions['ProductCategory'].value_counts())

print("\n--- Unique Regions ---")
print(transactions['Region'].unique())

print("\n--- Mean Amount Grouped by Region ---")
print(transactions.groupby('Region')['Amount'].mean())
```

```python
# 3. Manipulating the DataFrame
# Modify Amount for TransactionID 102 (matches index label 1 based on default indexing)
transactions.loc[transactions['TransactionID'] == 102, 'Amount'] = 165

# Add 'Discount' column (10% of Amount)
transactions['Discount'] = transactions['Amount'] * 0.10

# Remove row with TransactionID 109
transactions = transactions[transactions['TransactionID'] != 109]

# Delete 'Discount' column
transactions = transactions.drop(columns=['Discount'])

print("\n--- Final Manipulated DataFrame ---")
print(transactions)
```
